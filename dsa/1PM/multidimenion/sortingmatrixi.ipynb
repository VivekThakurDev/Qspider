# Reverse the 2D array

1) reverse each rows 
2) reverse each column
3) rotate 90 deg clockwise
4) rotate 90 deg anticlockwise
5) rotate 180 deg



```python
def spiralOrder(matrix):
    l=[]
    left=0
    right=len(matrix[0])-1
    top=0
    bottom=len(matrix)-1

    for i in range(left,right):
        i.append(matrix[top][i])

    top+=1

    for i in range (top,bottom):
        i.append(matrix[i][right])

    right -=1

    return 1


matrix=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
spiralOrder(matrix)
```
