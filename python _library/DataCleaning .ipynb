```python
### Filling NaN Value
# Syntax: var['col_name']= var['col_name'].fillna(value)
```

```python
df= pd.DataFrame(
  {
      'Ename':['veivek','veivek','kushi','sumit','chuccu',np.nan,np.nan],
      'Eno':[102,102,np.nan,105,106,104,np.nan],
      'Sal':['6LPA','5LPA',np.nan,np.nan,np.nan,'15LPA',np.nan]
      
  }
)
df
```

```python
df['PF']=[100,200,2200,np.nan,500,744,np.nan]
df
```

```python
d= round(df['PF'].mean())
```

```python
df['PF']= df['PF'].fillna(d)
```

```python
df
```

```python
df['Eno']= df['Eno'].fillna('Terminated')
```

```python
df
```

```python
##  Forward filling and backward filling 
# Forword filling
# Syntax:
# var['col_name']= var['col_name'].ffill()
```

```python
df['Sal']=df['Sal'].ffill()
```

```python
df
```

```python
##  Forward filling and backward filling 
# Backward filling
# Syntax:
# var['col_name']= var['col_name'].bfill()
```

```python
df
```

```python
df['Sal']= df['Sal'].bfill()
```

```python
df['PF']= df['PF'].fillna()
```

```python
df.info()
```

```python
# var['col_nmae']=var['Col_name'].astype(data_type)
```

```python
df['PF']= df['PF'].astype('int')
```
