```python
 # Filtering :- 
 ages=np.array([
     [21,17,29,20,10,30,18,65],
     [39,22,15,18,54,20,21,16]
            ])
```

```python
Bacha = ages[ages<18]
```

```python
Bacha
```

```python
chaca = ages[ages>18]
```

```python
chaca
```

```python
 budha = (ages<60) & (ages>18)
 budha
```

```python
budha_age_int = ages[(ages<60) & (ages>18)]
budha_age_int
```

```python
budha_age_int_OR = ages[(ages<60) | (ages>18)]
budha_age_int_OR
```

```python
budha_age_int_OR_bool = [(ages<60) | (ages>18)]
budha_age_int_OR_bool
```

```python
budha_age_int_NOT = ages[~((ages<60) | (ages>18))]
budha_age_int_NOT
```

```python
budha_age = ages[(ages<18) | (ages>60)]
budha_age
```
