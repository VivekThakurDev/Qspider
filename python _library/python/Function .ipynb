```python
#  Perfect Number :
def perfect(n):
    temp=n
    sum=0
    for i in range(1,n):
        if n%i==0:
            sum+=i
    if n== sum:
        return 'perfect'
    return 'not perfect'
print(perfect(28))
```

```python
# factorial
def fact(n):
    pro=1
    for i in range(1,n+1):
        pro*=i
    return pro
print(fact(3))
```

```python
def strong(n):
    to =0
    temp=n
    while temp!=0:
        product=1
        ld=temp%10
        for i in range(1,ld+1):
            product=product*i
        temp=temp//10
        to+=product

    if to == n:
        return 'Strong'
    return 'Not Strong'
print(strong(49))
```
