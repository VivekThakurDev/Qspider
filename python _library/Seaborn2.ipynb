#### Matrix plot:

to work with matrix plot we have be good with
1) pivot_table
2) corr()
- syntax:- var.pibot_table(values=[col...], index=['cat_col'],aggfunc=[fname,...])


matrix plot------>    heatmap()
- syntax:- sns.heatmap(data,annot=True)

```python
df= pd.read_csv('winequalityN.csv')
```

```python
df
```

```python
mean= df.pivot_table(values=['fixed acidity','citric acid','pH'],index=['quality'])
sns.heatmap(mean,annot=True,cmap='Reds')
plt.show()
```

```python
# corr():-   it find the correlation values
# var.corr()-----> it accept only numeric columns always corr value lies between -1 or 1
# 1) when corr value is nearly equal to +1 ---->. high corelated (directly)
# 2) when corr value is nearly equal to -1 ---->negatively correlated (inversely)
```

```python
corr_data= df.iloc[:,1:].corr()
corr_data
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(corr_data,annot=True,cmap='Reds')
```
