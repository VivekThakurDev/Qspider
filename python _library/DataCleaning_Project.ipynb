```python
price= {'price':'price($)'}
df.rename(columns=price,inplace=True)
```

```python
df['price($)'] = (
    df['price($)']
    .str.replace(',', '')
    .str.replace('$', '')
)

df['price($)'] = pd.to_numeric(df['price($)'], errors='coerce')
```

```python
df['discount']= (
               df['discount']
               .str.replace('-','')
               .str.replace('%','')
)
df['discount']=pd.to_numeric(df['discount'],errors='coerce')
                             
```
