## histplot
- It is used to visulize the distribution of data based on one column
- Syntax:-   sns.histplot(data=data_var,x='col_name')

```python
#
sns.histplot(data=tips,x='total_bill',color='red',bins=10, kde=True, hue='sex')
plt.show()
```

```python
sns.histplot(data=tips,x='total_bill',color='red',bins=10, kde=True, hue='sex')
plt.show()
```

#### Arguments of hist:- 
- bines= it will return the bine
- kde(kernal desity estimate):-  it will return the connecting line of data set  (kde=True/False)
- hue:- you have pass the data that you can categorize it    (hue='col_name')

```python
# palette :- it is use to change the  categorrie color
# Values for parameter palette
#Set1
#Set2
#Set3
# viridis(*)
# coolwarn
# inferna
#  magma
# Plasma
sns.histplot(data=tips,x='total_bill',color='red', kde=True, hue='sex',palette='Oranges')
plt.show()
```

```python
# Multiple :- it is use to target  the difference
# multiple='layer'/'stack'/'fill'/'dodge'
sns.histplot(data=tips,x='total_bill',color='red',bins=10, kde=True, hue='sex',palette='Oranges',multiple='dodge')
plt.show()

```

```python
# Kde
sns.histplot(data=tips,x='total_bill',color='red', kde=True)
plt.show()
```

```python
sns.histplot(data=tips,x='total_bill',color='red',bins=10, kde=True, hue='sex',multiple='stack')
plt.show()
```

### Boxplot():-
- It is used to identify outliers ,extremly low values or extremly high values
- syntax:- sns.boxplot(x='col_1',y='col_2',data=var)

```python
sns.boxplot(x='type',y='alcohol',data=df)
```

```python
Q1= df['quality'].quantile(0.25)
Q3=df['quality'].quantile(0.75)
Q1,Q3
```

```python
IQR=Q3-Q1  # IQR(Inter Quartile Range)
IQR
```

```python
lower_bound= IQR*1.5-Q1
upper_bound=IQR*1.5+Q3

lower_bound,upper_bound
```

```python
df1=df[(df['quality']> upper_bound) | (df['quality']< lower_bound)]
df1
```
