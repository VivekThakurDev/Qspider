```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('sales_data.csv', encoding='ISO-8859-1' )
```

```python
df.head(2)
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df[df['Amount'].isnull()].count()
```

```python
df.dropna(inplace=True)
```

```python
df[df.isnull()].count()
```

```python
df.info()
```

```python
df.rename(columns={'Marital_Status':'Marital'},inplace=True)
```

```python
df.head()
```

```python
# af= {0:'Unmarried',1:'Married'}
# df['Marital']=df['Marital'].map(af)
#
#or 
df['Marital']=df['Marital'].replace({0:'Single',1:'Married'})
```

```python
df.head()
```

```python
df['Gender']=df['Gender'].replace({'F':'Female','M':'Male'})
```

```python
df.head()
```

```python
# Visulization the count of males and females based on gender column using seaborn.
a=sns.countplot(x='Gender',data=df,hue='Gender')
for i in a.containers:
    a.bar_label(i)
```

```python
# total amount spend by gender.
a= df.groupby('Gender')['Amount'].sum()
a=pd.DataFrame(a)

b= sns.barplot(x='Gender',y='Amount',data=a,hue='Gender')
for i in b.containers:
    b.bar_label(i)
    
```

```python
# count of each age group
c= sns.countplot(x='Age Group',data=df,hue='Age Group')
for i in c.containers:
    c.bar_label(i)
```

```python
#  AMount spend by each age group
a= df.groupby('Age Group')['Amount'].sum()
a=pd.DataFrame(a)
plt.figure(figsize=(20,10))
d= sns.barplot(x='Age Group',y='Amount',data=a,hue='Age Group')
for i in d.containers:
    d.bar_label(i)
```

```python

plt.figure(figsize=(20,10))
v= sns.countplot(x='State',data=df,hue='State')
plt.xticks(rotation=45)
for i in v.containers:
    v.bar_label(i)
```
